```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("img.png")
if image is None:
    raise FileNotFoundError("Could not load img.png")
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
width = int(input("Enter border width in pixels: "))
if width < 0:
    raise ValueError("Border width cannot be negative")
value = int(input("Enter border intensity (0–255): "))
if not 0 <= value <= 255:
    raise ValueError("Border intensity must be from 0 to 255")
binary = (gray >= 128).astype(np.uint8) * 255
gray_padded = cv2.copyMakeBorder(gray, width, width, width, width, cv2.BORDER_CONSTANT, value=value)
binary_padded = cv2.copyMakeBorder(binary, width, width, width, width, cv2.BORDER_CONSTANT, value=value)
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(gray_padded, cmap="gray", vmin=0, vmax=255); ax[0].set_title("Grayscale with border")
ax[1].imshow(binary_padded, cmap="gray", vmin=0, vmax=255); ax[1].set_title("Binary with border")
for a in ax: a.axis("off")
plt.show()
```

```python
gray_complement = 255 - gray
binary_complement = 255 - binary
fig, ax = plt.subplots(2, 2, figsize=(9, 7))
for a, im, title in zip(ax.flat, [gray, gray_complement, binary, binary_complement], ["Gray", "Gray complement", "Binary", "Binary complement"]):
    a.imshow(im, cmap="gray", vmin=0, vmax=255); a.set_title(title); a.axis("off")
plt.tight_layout(); plt.show()
```
