```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("img.png")
if image is None:
    raise FileNotFoundError("Could not load img.png")
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
threshold_mean = np.mean(gray)
binary_mean = (gray >= threshold_mean).astype(np.uint8) * 255
print("Mean threshold:", threshold_mean)
plt.imshow(binary_mean, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

```python
threshold = int(input("Enter threshold (0–255): "))
if not 0 <= threshold <= 255:
    raise ValueError("Threshold must be from 0 to 255")
binary_user = (gray >= threshold).astype(np.uint8) * 255
plt.imshow(binary_user, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```
