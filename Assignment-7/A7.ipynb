```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("img.png")
if image is None:
    raise FileNotFoundError("Could not load img.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
image_float = image.astype(np.float32)
```

```python
strength = float(input("Enter sharpening strength (e.g. 1.0): "))
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)
laplacian_sharp = np.clip(image_float - strength * laplacian, 0, 255).astype(np.uint8)
```

```python
sobel_x = cv2.Sobel(gray, cv2.CV_32F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_32F, 0, 1, ksize=3)
sobel_edges = cv2.magnitude(sobel_x, sobel_y)
sobel_edges = np.clip(sobel_edges, 0, 255)
sobel_sharp = np.clip(image_float + strength * sobel_edges[:, :, None], 0, 255).astype(np.uint8)
```

```python
low_threshold = int(input("Enter Canny lower threshold: "))
high_threshold = int(input("Enter Canny upper threshold: "))
if not 0 <= low_threshold <= high_threshold <= 255:
    raise ValueError("Thresholds must satisfy 0 <= lower <= upper <= 255")
canny_edges = cv2.Canny(gray, low_threshold, high_threshold)
canny_sharp = np.clip(image_float + strength * canny_edges[:, :, None], 0, 255).astype(np.uint8)
```

```python
blurred = cv2.GaussianBlur(image, (5, 5), sigmaX=3, sigmaY=3).astype(np.float32)
amount = float(input("Enter unsharp masking amount (e.g. 1.0): "))
unsharp = np.clip(image_float + amount * (image_float - blurred), 0, 255).astype(np.uint8)
```

```python
results = [("Original", image), ("Laplacian", laplacian_sharp), ("Sobel", sobel_sharp), ("Canny", canny_sharp), ("Unsharp masking", unsharp)]
fig, axes = plt.subplots(1, 5, figsize=(18, 5))
for ax, (title, result) in zip(axes, results):
    ax.imshow(result)
    ax.set_title(title)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
