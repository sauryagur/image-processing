```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("img.png")
if image is None:
    raise FileNotFoundError("Could not load img.png")
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
gray_mean = np.mean(rgb, axis=2).astype(np.uint8)
plt.imshow(gray_mean, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

```python
weights = np.array([float(input(f"Enter {c} weight (0–1): ")) for c in "RGB"])
if np.any(weights < 0) or np.any(weights > 1) or not np.isclose(weights.sum(), 1):
    raise ValueError("Weights must be between 0 and 1 and sum to 1")
gray_weighted = np.dot(rgb, weights).clip(0, 255).astype(np.uint8)
plt.imshow(gray_weighted, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```
