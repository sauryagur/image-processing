```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("img.png")
if image is None:
    raise FileNotFoundError("Could not load img.png")
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
gray_float = gray.astype(np.float32)
log_image = (255 / np.log1p(255)) * np.log1p(gray_float)
log_image = log_image.astype(np.uint8)
plt.imshow(log_image, cmap="gray", vmin=0, vmax=255); plt.axis("off"); plt.show()
```

```python
gamma = float(input("Enter gamma (> 0): "))
if gamma <= 0:
    raise ValueError("Gamma must be greater than 0")
gamma_image = (255 * (gray_float / 255) ** gamma).astype(np.uint8)
plt.imshow(gamma_image, cmap="gray", vmin=0, vmax=255); plt.axis("off"); plt.show()
```

```python
in_low = int(input("Input range lower bound: "))
in_high = int(input("Input range upper bound: "))
out_low = int(input("Output range lower bound: "))
out_high = int(input("Output range upper bound: "))
if not (0 <= in_low < in_high <= 255 and 0 <= out_low <= 255 and 0 <= out_high <= 255):
    raise ValueError("Use intensity values from 0 to 255 and increasing input bounds")
sliced = out_low + (gray_float - in_low) * (out_high - out_low) / (in_high - in_low)
sliced = np.clip(sliced, min(out_low, out_high), max(out_low, out_high)).astype(np.uint8)
plt.imshow(sliced, cmap="gray", vmin=0, vmax=255); plt.axis("off"); plt.show()
```
