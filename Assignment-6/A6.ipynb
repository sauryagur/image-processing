```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("img.png")
if image is None:
    raise FileNotFoundError("Could not load img.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
size = int(input("Enter odd filter size: "))
if size < 1 or size % 2 == 0:
    raise ValueError("Filter size must be a positive odd number")
padding = input("Padding (zero, replication, reflect): ").lower()
border_types = {"zero": cv2.BORDER_CONSTANT, "replication": cv2.BORDER_REPLICATE, "reflect": cv2.BORDER_REFLECT}
if padding not in border_types:
    raise ValueError("Choose zero, replication, or reflect")
border_type = border_types[padding]
```

```python
average = cv2.blur(image, (size, size), borderType=border_type)
```

```python
radius = size // 2
coordinates = np.arange(-radius, radius + 1)
x, y = np.meshgrid(coordinates, coordinates)
weights = np.maximum(np.abs(x), np.abs(y)).astype(np.float32)
weights /= weights.sum()
weighted = cv2.filter2D(image, -1, weights, borderType=border_type)
```

```python
sigma = float(input("Enter Gaussian sigma (> 0): "))
if sigma <= 0:
    raise ValueError("Sigma must be greater than 0")
g = cv2.getGaussianKernel(size, sigma)
gaussian_kernel = g @ g.T
gaussian = cv2.filter2D(image, -1, gaussian_kernel, borderType=border_type)
```

```python
results = [("Original", image), ("Averaging", average), ("Weighted", weighted), ("Gaussian", gaussian)]
fig, axes = plt.subplots(1, 4, figsize=(15, 5))
for ax, (title, result) in zip(axes, results):
    ax.imshow(result)
    ax.set_title(title)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
